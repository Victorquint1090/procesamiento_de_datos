# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.hist(data.iloc[:, 0], bins=20, alpha=0.5, label='Original')
plt.hist(data_minmax.iloc[:, 0], bins=20, alpha=0.5, label='MinMax')

plt.xlabel(data.columns[0])
plt.ylabel('Frecuencia')
plt.title('Comparación antes y después de MinMaxScaler')
plt.legend()
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
scaler = StandardScaler()

data_scaled = data.copy()

data_scaled[['sepal length (cm)', 'sepal width (cm)']] = scaler.fit_transform(
    data[['sepal length (cm)', 'sepal width (cm)']]
)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    data_scaled['sepal length (cm)'],
    data_scaled['sepal width (cm)']
)

plt.xlabel('Sepal Length (escalado)')
plt.ylabel('Sepal Width (escalado)')
plt.title('Sepal Length vs Sepal Width - StandardScaler')

plt.show()
```

### 4. ¿Qué técnica de escalado crees que es más adecuada para este conjunto de datos? ¿Por qué?

```python
Considero que StandardScaler es la técnica más adecuada para este conjunto de datos, porque estandariza las variables haciendo que tengan una media cercana a 0 y una desviación estándar de 1. Esto permite comparar las características en una escala común sin que alguna variable tenga mayor influencia debido a sus unidades o magnitud. Además, es una buena opción para algoritmos sensibles a la escala, como KNN, SVM y regresión.
```
